NumPy Arrays and Array Operations

```python
arr1 = np.array([10, 20, 30, 40])   
arr2 = np.zeros((2,3))              
arr3 = np.ones(5)                   
arr4 = np.arange(0, 10, 2)         
arr5 = np.linspace(0, 1, 5)         

arr1, arr2, arr3, arr4, arr5
```

```python
print("Shape:", arr4.shape)
print("Dimensions:", arr4.ndim)
print("Data type:", arr4.dtype)
print("Size:", arr4.size)
print("Item size:", arr4.itemsize)
```

shape → number of rows and columns.

ndim → number of dimensions.

dtype → type of elements (int, float, etc.).

size → total number of elements.

itemsize → memory used per element.

Indexing & Slicing

```python
print(arr1[2])       # single element
print(arr4[1:4])     # slice
print(arr2[0,1])     # 2D indexing
print(arr2[:,1])     # column slice
```

Indexing allows access to specific elements, while slicing extracts ranges. In 2D arrays, you use row and column indices.

Reshape & Transpose

```python
arr6 = np.arange(12).reshape(3,4)
print("Reshaped:\n", arr6)

print("Flatten:", arr6.flatten())
print("Ravel:", arr6.ravel())
print("Transpose:\n", arr6.T)
```

reshape() changes dimensions.

flatten() returns a copy as 1D.

ravel() returns a view (more memory-efficient).

transpose() flips rows and columns.

Element-wise Operations

```python
print(arr4 + 5)
print(arr4 * 2)
print(np.sqrt(arr4))
```

NumPy supports vectorized operations — no need for loops. This makes computations faster and more readable.

Advanced NumPy for EDA

Broadcasting-automatically expands smaller arrays to match larger ones, enabling operations without explicit loops.

```python
arr7 = np.ones((3,3))
arr8 = np.array([1,2,3])
print(arr7 + arr8)
```

Aggregation functions summarize data — useful for statistical analysis.

```python
print("Sum:", arr6.sum())
print("Mean:", arr6.mean())
print("Std:", arr6.std())
```

Axis-based operations let you compute summaries across rows (axis=1) or columns (axis=0).

```python
print("Row-wise sum:", arr6.sum(axis=1))
print("Column-wise sum:", arr6.sum(axis=0))
```

Boolean indexing filters arrays based on conditions — very useful for selecting subsets of data.

```python
scores = np.array([55, 72, 88, 91, 45])
print("Scores > 70:", scores[scores > 70])
```

Fancy indexing allows selection of specific positions, not just ranges.

```python
indices = [0,2,4]
print("Selected scores:", scores[indices])
```

Random Number Generation-Using a fixed seed ensures reproducibility — the same random numbers are generated every time.

```python
np.random.seed(42)
rand_arr = np.random.randint(1,100, size=(5,))
print(rand_arr)
```

Exploratory Visualisation

Numerical variable

```python
plt.figure(figsize=(6,4))
sns.histplot(data=df, x="G3", bins=10, kde=True, color="skyblue")
plt.title("Distribution of Final Grade (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.show()
```

This histogram shows how final grades (G3) are distributed among students. The KDE curve highlights the overall shape of the distribution.

Observations:

Most students scored between 5–15.

A few students scored very low (below 5), indicating weak performance.

The distribution is slightly skewed, with more students achieving mid-range scores than extremes.

Categorial Variable Exploration

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x="sex", palette="Set2")
plt.title("Gender Distribution of Students")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.show()
```

This bar chart shows the gender distribution (sex) in the dataset.

Observations:

The dataset contains more female students than male students.

Gender balance is not perfectly equal, which may influence comparative analysis.

Relationship Exploration

```python
plt.figure(figsize=(6,4))
sns.scatterplot(data=df, x="studytime", y="G3", hue="sex", palette="coolwarm")
plt.title("Study Time vs Final Grade by Gender")
plt.xlabel("Study Time (hours/week)")
plt.ylabel("Final Grade (G3)")
plt.legend(title="Gender")
plt.show()
```

This scatter plot explores the relationship between study time (studytime) and final grades (G3), with gender as a categorical distinction.

Observations:

Students with higher study time generally achieve better grades.

The relationship is positive but not perfectly linear — some students with high study time still score lower.

Gender differences appear minimal, suggesting study time is a stronger factor than gender.

EDA Communication and Findings

Key Patterns
Students with higher study time generally achieve better final grades (G3).

Students with more absences tend to score lower, showing a negative relationship between attendance and performance.

The dataset highlights that behavioral factors (studytime, attendance) are stronger indicators of performance than demographic variables.

Answering the Analytical Question
The central question was: “How do study time and attendance relate to students' final performance?”

Evidence shows that consistent study habits and regular attendance are linked to higher grades.

Students who study more and attend regularly perform better, while frequent absences reduce performance outcomes.

Observations vs Assumptions
Observation: Higher study time correlates with higher grades.

Observation: More absences correlate with lower grades.

Assumption to avoid: Claiming study time or attendance causes performance changes — correlation does not prove causation. Other hidden factors (motivation, teaching quality, peer support) may also influence outcomes.

Limitations:
Dataset is limited to one region/context and may not generalize globally.

Attendance is measured only as raw absences, not quality of participation.

Study time is self-reported and may not reflect effective study habits.

Other important variables (like teaching quality, peer influence, or mental health) are not included.

Follow-up Analyses
Explore interaction effects: Does high study time offset the negative impact of absences?

Compare early grades (G1, G2) with final grade (G3) to see if attendance/study habits consistently influence performance over time.

Investigate lifestyle factors (like Dalc/Walc alcohol consumption) alongside attendance to see if they compound performance outcomes.

Conclusion:
The analysis answers the central question: study time and attendance both strongly relate to final performance.

Students who study more and attend regularly tend to achieve higher grades.

While the patterns are clear, causation cannot be claimed — further controlled studies would be needed.

Still, the evidence highlights the importance of consistent study habits and attendance in academic success.
